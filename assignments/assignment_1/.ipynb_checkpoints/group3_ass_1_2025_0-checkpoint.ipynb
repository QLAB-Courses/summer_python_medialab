a.Declare one variable for each one these types: : int , float , string , boolean.
- Print  the varible's type using  type () function




````python
# Declaramos variables de distintos tipos
num_entero = 10        # int
num_decimal = 3.14     # float
texto = "Hola"         # str
es_verdadero = True    # bool

# Imprimimos el tipo de cada variable
print(type(num_entero))   # <class 'int'>
print(type(num_decimal))  # <class 'float'>
print(type(texto))        # <class 'str'>
print(type(es_verdadero)) # <class 'bool'>
````

g.1.Add tree additional courses in 3 different ways( one for each way):
   - Sum (+) a list that contains an additional course. 
   - Use append. 
   - Use extend. 

````python
courses = ["History", "Economy", "Anthropology", "Sociology", "Literature"]

# Usando el operador + para sumar una lista con un curso adicional
courses = courses + ["Philosophy"]

# Usando append() para agregar un curso más
courses.append("Teology")

# Usando extend() para agregar otro curso
courses.extend(["Political Science"])

print(courses)

````

h.Create a dictionary that representst a fruits inventory:
 - Add at least 5 fruits to the inventory
- Print the amount of apples in stock 
- Modify the amount of grapes


````python
# Crear el inventario de frutas con 5 frutas y sus cantidades
fruit_inventory = {
    "manzanas": 20,
    "bananas": 15,
    "naranja": 30,
    "uvas": 25,
    "mango": 10
}

# Imprimir la cantidad de manzanas en stock
print("manzanas in stock:", fruit_inventory["manzanas"])

# Modificar la cantidad de uvas
fruit_inventory["uvas"] = 40  # Cambiar la cantidad de uvas a 40

# Imprimir el inventario actualizado
print("Updated inventory:", fruit_inventory)
````

i.  Write a dictionary with the local GMT of the nexts cities:
- Antananarivo, Madagascar, 
- Beirut, Lebanon, 
- Jakarta, Indonesia, 
- Teherán, Iran, 
- Vientiane, Laos, 



Hint: Lima would be: `"Lima": -5`

````python
gmt_cities = {
    "Antananarivo": 3,  # Madagascar (GMT+3)
    "Beirut": 2,        # Lebanon (GMT+2, puede ser +3 en horario de verano)
    "Jakarta": 7,       # Indonesia (GMT+7)
    "Teherán": 3.5,     # Irán (GMT+3:30)
    "Vientiane": 7      # Laos (GMT+7)
}

# Imprimir el diccionario
print(gmt_cities)
````

 j.Add the nexts cities to the dictionary:
- Astana, Kazakhstan,

````python

# Agregar las nuevas ciudades al diccionario
gmt_cities["Astana"] = 6    # Kazajistán (GMT+6)
gmt_cities["Uagadugú"] = 0   # Burkina Faso (GMT+0)
gmt_cities["Balmopan"] = -6  # Belice (GMT-6)

# Imprimir el diccionario actualizado
print(gmt_cities)
````

m.Sum two arrays and then multiply the output by an scalar

````python
import numpy as np

# Creamos dos arrays de ejemplo
array1 = np.array([1, 2, 3, 4, 5])
array2 = np.array([5, 4, 3, 2, 1])

# Sumamos los arrays
sum_result = array1 + array2  

# Multiplicamos el resultado por un escalar (ejemplo: 3)
scalar = 3
final_result = sum_result * scalar

print(final_result)
````
