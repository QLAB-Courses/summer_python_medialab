```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
from io import StringIO
import os
import time

# Definir la ruta donde se guardarán los archivos
path_siaf = r'C:\Users\Miguel\Documents\5. 2025\Extracción y análisis de Datos - Media QLab\SIAF'
path_siaf_regional = os.path.join(path_siaf, 'regional')

# Crear la carpeta si no existe
os.makedirs(path_siaf_regional, exist_ok=True)

# Iniciar el navegador fuera del bucle
driver = webdriver.Chrome()

# Loop para iterar sobre los años 2014 - 2024
for year in range(2014, 2025):
    print(f"📊 Procesando año {year}...")

    url = f"https://apps5.mineco.gob.pe/transparencia/Navegador/default.aspx?y={year}&ap=ActProy"
    driver.get(url)

    try:
        # Esperar a que cargue el frame y cambiar a él
        WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.ID, "frame0")))
        frame = driver.find_element(By.ID, "frame0")
        driver.switch_to.frame(frame)

        # Click en "Niveles de Gobierno"
        WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.XPATH, '//*[@id="ctl00_CPH1_BtnTipoGobierno"]'))).click()

        # Click en "Gobiernos Regionales"
        WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.XPATH, '//*[@id="ctl00_CPH1_RptData_ctl03_TD0"]/input'))).click()

        # Click en "Sector"
        WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.XPATH, '//*[@id="ctl00_CPH1_BtnSector"]'))).click()

        # Intentar seleccionar el botón de gobiernos regionales
        try:
            regional_governments_button = WebDriverWait(driver, 10).until(
                EC.element_to_be_clickable((By.XPATH, '//*[@id="ctl00_CPH1_RptData_ctl02_TD0"]/input'))
            )
            driver.execute_script("arguments[0].click();", regional_governments_button)
            print(f"✅ Gobierno Regional seleccionado para {year}")
        except:
            print(f"⚠️ No se encontró el botón de Gobierno Regional en {year}")

        # Click en "Pliego"
        WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.XPATH, '//*[@id="ctl00_CPH1_BtnPliego"]'))).click()

        # Esperar y extraer la tabla
        WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.XPATH, "//table[@class='Data']")))
        table_element = driver.find_element(By.XPATH, "//table[@class='Data']")
        table_html = table_element.get_attribute('outerHTML')
        table_html_io = StringIO(table_html)
        table_df = pd.read_html(table_html_io)[0]

        # Obtener encabezados
        headers_r0 = [e.text for e in driver.find_element(By.XPATH, '//*[@id="ctl00_CPH1_Mt0_Row0"]').find_elements(By.TAG_NAME, 'td')]
        headers_r1 = [e.text for e in driver.find_element(By.XPATH, '//*[@id="ctl00_CPH1_Mt0_Row1"]').find_elements(By.TAG_NAME, 'td')]

        headers_r0 = [h for h in headers_r0 if h not in ['Ejecución', 'Avance % ']]
        headers = headers_r0 + headers_r1 + ['Avance %']

        table_df.columns = headers
        table_df = table_df.drop(table_df.columns[0], axis=1)

        # Guardar en Excel
        table_path = os.path.join(path_siaf_regional, f'regional_table_{year}.xlsx')
        table_df.to_excel(table_path, index=False)
        print(f"📂 Archivo guardado: {table_path}")

    except Exception as e:
        print(f"❌ Error en el año {year}: {e}")

# Cerrar el navegador
driver.quit()
print("✅ Proceso completado para todos los años.")
```

```python
!pip install xlsxwriter

# Ruta donde están los archivos de cada año
path_siaf_regional = r'C:\Users\Miguel\Documents\5. 2025\Extracción y análisis de Datos - Media QLab\SIAF\regional'

# Ruta del archivo consolidado
excel_file_path = os.path.join(path_siaf_regional, 'SIAF_2014_2024.xlsx')

# Obtener lista de archivos Excel en la carpeta
excel_files = [f for f in os.listdir(path_siaf_regional) if f.endswith('.xlsx')]

# Crear el archivo consolidado
with pd.ExcelWriter(excel_file_path, engine='xlsxwriter') as writer:
    for file in excel_files:
        file_path = os.path.join(path_siaf_regional, file)

        # Extraer el año del nombre del archivo (suponiendo que el año está en el nombre, ej: "SIAF_2014.xlsx")
        year = ''.join(filter(str.isdigit, file))  # Extrae los números del nombre del archivo

        if year:  # Si se pudo extraer un año
            print(f"Procesando {file} → Año {year}")

            # Leer el archivo Excel
            df = pd.read_excel(file_path)

            # Guardar en la hoja correspondiente del nuevo archivo consolidado
            df.to_excel(writer, sheet_name=str(year), index=False)

print(f"Archivo consolidado guardado en: {excel_file_path}")
```
