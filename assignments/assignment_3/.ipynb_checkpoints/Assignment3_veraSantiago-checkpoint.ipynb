```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

# Inicializar el WebDriver
def iniciar_driver():
    driver = webdriver.Edge()
    driver.maximize_window()
    return driver

# Acceder a la página web
def acceder_pagina(driver, url):
    driver.get(url)
    time.sleep(5)  # Esperar a que la página cargue completamente

# Cambiar al frame principal
def cambiar_frame(driver, frame_id):
    print("Cambiando al frame principal...")
    driver.switch_to.frame(frame_id)

# Intentar hacer clic en el radio button
def seleccionar_nivel_gobierno(driver):
    print("Buscando elementos en el frame...")
    wait = WebDriverWait(driver, 10)
    
    try:
        radio_button = wait.until(
            EC.element_to_be_clickable((By.CSS_SELECTOR, "input[type='radio']"))
        )
        print("Elemento encontrado, intentando hacer clic...")
        radio_button.click()
        print("Clic realizado con éxito")
    except Exception:
        print("No se pudo encontrar el elemento del radio button")
        radio_buttons = driver.find_elements(By.CSS_SELECTOR, "input[type='radio']")
        print(f"Número de radio buttons encontrados: {len(radio_buttons)}")
        for rb in radio_buttons:
            print(f"ID: {rb.get_attribute('id')}, Name: {rb.get_attribute('name')}")

# Función principal
def main():
    url = "https://apps5.mineco.gob.pe/transparencia/Navegador/default.aspx?y=2014&ap=ActProy"
    driver = iniciar_driver()
    
    try:
        acceder_pagina(driver, url)
        cambiar_frame(driver, "frame0")
        seleccionar_nivel_gobierno(driver)
    except Exception as e:
        print(f"Error: {str(e)}")
    finally:
        time.sleep(5)
        driver.quit()

if __name__ == "__main__":
    main()
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

# Iniciar el WebDriver
driver = webdriver.Edge()

try:
    # Acceder a la página web
    url = "https://apps5.mineco.gob.pe/transparencia/Navegador/default.aspx?y=2014&ap=ActProy"
    driver.get(url)
    driver.maximize_window()

    # Esperar a que la página cargue completamente
    wait = WebDriverWait(driver, 10)

    # Seleccionar el botón de nivel de gobierno regional
    regional_governments_level_button = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="ctl00_CPH1_RptData_ctl03_TD0"]/input'))
    )
    regional_governments_level_button.click()
    
    # Esperar antes de proceder
    time.sleep(2)

    # Hacer clic en el botón de sector
    sector_button = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="ctl00_CPH1_BtnSector"]'))
    )
    sector_button.click()

    time.sleep(2)

    # Seleccionar el botón de gobiernos regionales
    regional_goverments_button = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="ctl00_CPH1_RptData_ctl02_TD0"]/input'))
    )
    regional_goverments_button.click()

    time.sleep(2)

    # Hacer clic en el botón de pliego
    pliego_button = wait.until(
        EC.element_to_be_clickable((By.XPATH, '//*[@id="ctl00_CPH1_BtnPliego"]'))
    )
    pliego_button.click()

except Exception as e:
    print(f"Error: {str(e)}")

finally:
    time.sleep(5)
    driver.quit()
```
