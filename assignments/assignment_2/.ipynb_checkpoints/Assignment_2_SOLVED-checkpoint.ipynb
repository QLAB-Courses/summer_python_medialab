1. Calculate the Average Trip Distance per Month

```python
# Group by pickup_month and calculate the mean trip_distance
avg_distance_per_month = df.groupby('pickup_month')['trip_distance'].mean()
    #Groups the dataset by month.
    #Computes the average trip distance for each month.
```

```python
# Display the result
print(avg_distance_per_month)
```

2. Count the Total Trips from JFK for Each Day of the Month

```python
# Filter trips that started at JFK (pickup_location_code == 2)
jfk_trips = df[df['pickup_location_code'] == 2]
    # Filters rows where pickup_location_code is JFK.
```

```python
# Group by pickup_day and count the number of trips
trips_from_jfk_per_day = jfk_trips.groupby('pickup_day').size()
    # Groups the filtered dataset by day of the month.
    # Counts the number of trips per day.
```

```python
# Display the result
print(trips_from_jfk_per_day)
```

3. Calculate the Average Speed by Pickup Day of the Week

To calculate speed, we use the formula:

\[
\text{Speed} = \frac{\text{Distance}}{\text{Time (in hours)}}
\]

We must convert `trip_length` (seconds) into hours.


```python
# Convert trip_length from seconds to hours
df['trip_length_hours'] = df['trip_length'] / 3600
    
```

```python
# Calculate speed in miles per hour (mph)
df['speed_mph'] = df['trip_distance'] / df['trip_length_hours']
```

```python
# Group by pickup_dayofweek and calculate the mean speed
avg_speed_per_dayofweek = df.groupby('pickup_dayofweek')['speed_mph'].mean()
```

```python
# Display the result
print(avg_speed_per_dayofweek)
```

## Part 2 Loops- For loops , while loops

a. Using a  “for loop” and the conditional “if else” for a range bettewn 1 to 100, evaluate "x" so that: [2 pts.]

- If x is a multiple of 3 print  “Fizz”
- If x is a multiple of 5 print  “Buzz”
- If x is a multiple of 3 and 5 print “FizzBuzz” 

```python
# Loop from 1 to 100
for x in range(1, 101):  
    # Check if x is divisible by both 3 and 5 first (most restrictive condition)
    if x % 3 == 0 and x % 5 == 0:
        print("FizzBuzz")
    # Check if x is only divisible by 3
    elif x % 3 == 0:
        print("Fizz")
    # Check if x is only divisible by 5
    elif x % 5 == 0:
        print("Buzz")
    # If none of the above, print the number itself
    else:
        print(x)
```

Using a while loop:
- b.9. Accumulate the total income (Total) day by day (adds sales totals by date). Stop the loop when the accumulated revenue exceeds $1000 and print the corresponding day. [0.5 pts]

```python
# Sort by Date
sales_df_sorted = sales_df.sort_values(by="Date")
```

```python
# Initialize variables
accumulated_revenue = 0
index = 0
```

```python
# While loop to accumulate revenue by date
while accumulated_revenue <= 1000 and index < len(sales_df_sorted):
    accumulated_revenue += sales_df_sorted.iloc[index]["Total"]
    index += 1

# Print the date when revenue exceeds $1000
print(f"Revenue exceeded $1000 on: {sales_df_sorted.iloc[index-1]['Date']}")
```

- b.10. Accumulate sales totals row by row. Stop the loop when the accumulated revenue exceeds $2000 and print how many sales were necessary. [0.5 pts]

```python
# Initialize variables
accumulated_revenue = 0
transaction_count = 0
```

```python
# While loop to accumulate revenue per transaction
while accumulated_revenue <= 2000 and transaction_count < len(sales_df):
    accumulated_revenue += sales_df.iloc[transaction_count]["Total"]
    transaction_count += 1

print(f"It took {transaction_count} transactions to exceed $2000.")
```

- b.11. Use a combination of `for and while` loops to calculate a rolling cumulative revenue: [0.5 pts]

For each category, use a for loop to iterate over its transactions.
Use a while loop to stop accumulating revenue for a category once it exceeds $ 3000.

Print the cumulative revenue for each category and the number of transactions it took to exceed $3000.

```python
# Group sales by Category
category_groups = sales_df.groupby("Category")
```

```python
# Loop through each category
for category, group in category_groups:
    accumulated_revenue = 0
    transaction_count = 0
    
    # While loop to accumulate revenue per category
    while accumulated_revenue <= 3000 and transaction_count < len(group):
        accumulated_revenue += group.iloc[transaction_count]["Total"]
        transaction_count += 1

    print(f"{category}: ${accumulated_revenue:.2f} in {transaction_count} transactions")
```

- b.12. Use a `nested for` loop to: [0.5 pts]

Group the sales by year and by date.

For each year, determine the date with the highest total revenue and the revenue value.
This task requires grouping and iterating within groups.

```python
# Convert Date to datetime type
sales_df["Date"] = pd.to_datetime(sales_df["Date"])

# Extract Year
sales_df["Year"] = sales_df["Date"].dt.year

# Group by Year
year_groups = sales_df.groupby("Year")
```

```python
# Iterate through each year
for year, group in year_groups:
    max_revenue_date = group.groupby("Date")["Total"].sum().idxmax()
    max_revenue_value = group.groupby("Date")["Total"].sum().max()
    
    print(f"In {year}, the highest revenue date was {max_revenue_date} with ${max_revenue_value:.2f}")
```

## Part 3  Try, continue and catch exceptions

In this case you will work with a `students_grade.csv` dataset. 

1. You are tasked with calculating the average grade for each student in the dataset. However, the dataset has some issues: [2 pts]

- Some scores are marked as "missing" or "invalid" (strings instead of numeric values).
- Some students may have incomplete grades across subjects.

To calculate the average:

Skip any "invalid" or "missing" values during calculations.
If a student has no valid grades, print a warning message for that student.


Use `try` and `except` to handle potential errors.
Use `continue` to skip invalid rows or scores.
Print the results in a clear format, showing the average grade for each student or a warning if no grades are available.

```python
import csv
from collections import defaultdict
```

```python
#Defining a Function to read data

def read_csv(file_path):
    """Reads the CSV file and returns a list of student records."""
    students = []
    with open(file_path, newline='', encoding='utf-8') as csvfile:
        reader = csv.DictReader(csvfile)
        for row in reader:
            students.append(row)
    return students
```

```python
# Question 1: Calculate the average grade per student

def calculate_student_averages(students):
    """Calculates the average grade per student, skipping invalid or missing values."""
    student_averages = {}
    
    for student in students:
        grades = []
        for subject in ['Math', 'English', 'Science', 'History']:
            try:
                grade = student[subject]
                if grade.lower() in ["missing", "invalid"]:
                    continue
                grades.append(float(grade))
            except (ValueError, KeyError, AttributeError):
                continue
        
        if grades:
            student_averages[student['Name']] = sum(grades) / len(grades)
        else:
            print(f"Warning: {student['Name']} has no valid grades.")
    
    return student_averages
```

```python
def main():
    file_path = r"C:\\Users\\Michael Encalada\\Documents\\GitHub\\QLAB\\summer_python_medialab\\assignments\\assignment_2\\students_grade.csv"
    students = read_csv(file_path)
    
    # Question 1: Student Averages
    student_averages = calculate_student_averages(students)
    print("\nStudent Averages:")
    for name, avg in student_averages.items():
        print(f"{name}: {avg:.2f}")

if __name__ == "__main__":
    main()
```

2. Calculate the average grade per subject for each year.  [2 pts]

Handle invalid or missing values during the calculation by skipping them.
Print the averages in a clean format (e.g., table or grouped output).

```python
# Question 2: Calculate the average grade per subject for each year

def calculate_subject_averages_by_year(students):
    """Calculates the average grade per subject for each year, skipping invalid or missing values."""
    yearly_subject_averages = {}
    
    for student in students:
        year = student['Year']
        if year not in yearly_subject_averages:
            yearly_subject_averages[year] = {subject: [] for subject in ['Math', 'English', 'Science', 'History']}
        
        for subject in ['Math', 'English', 'Science', 'History']:
            try:
                grade = student[subject]
                if grade.lower() in ["missing", "invalid"]:
                    continue  # Skip invalid/missing grades
                yearly_subject_averages[year][subject].append(float(grade))
            except (ValueError, KeyError, AttributeError):
                continue
    
    # Compute the averages
    formatted_data = []
    for year, subjects in sorted(yearly_subject_averages.items()):
        avg_row = [year]
        for subject in ['Math', 'English', 'Science', 'History']:
            grades = subjects[subject]
            avg = sum(grades) / len(grades) if grades else None
            avg_row.append(avg)
        formatted_data.append(avg_row)
    
    return formatted_data
```

```python
def main():
    file_path = r"C:\\Users\\Michael Encalada\\Documents\\GitHub\\QLAB\\summer_python_medialab\\assignments\\assignment_2\\students_grade.csv"
    students = read_csv(file_path)
    
    # Question 2: Subject Averages by Year
    subject_averages_by_year = calculate_subject_averages_by_year(students)
    
    df = pd.DataFrame(subject_averages_by_year, columns=["Year", "Math", "English", "Science", "History"])
    df.set_index("Year", inplace=True)
    print("\nSubject Averages by Year:")
    print(df.to_string(float_format="%.6f"))

if __name__ == "__main__":
    main()
```

3. Identify students who had an average grade of 90 or above across all years (excluding invalid/missing values).  [2 pts]

If a student has no valid grades, skip them with a warning.

```python
# Question 3: Identify students with an average grade of 90 or above
def identify_top_students(students):
    """Identifies students who have an average grade of 90 or above, skipping invalid or missing values."""
    top_students = []
    
    for student in students:
        student_id = student['Student_ID']
        name = student['Name']
        valid_grades = []
        
        for subject in ['Math', 'English', 'Science', 'History']:
            try:
                grade = student[subject]
                if grade.lower() in ["missing", "invalid"]:
                    continue  # Skip invalid/missing grades
                valid_grades.append(float(grade))
            except (ValueError, KeyError, AttributeError):
                continue
        
        if valid_grades:
            avg_grade = sum(valid_grades) / len(valid_grades)
            if avg_grade >= 90:
                top_students.append((student_id, name, avg_grade))
        else:
            print(f"Warning: {name} has no valid grades and is skipped.")
    
    # Sort by highest average grade
    top_students.sort(key=lambda x: x[2], reverse=True)
    return top_students
```

```python
def main():
    file_path = r"C:\\Users\\Michael Encalada\\Documents\\GitHub\\QLAB\\summer_python_medialab\\assignments\\assignment_2\\students_grade.csv"
    students = read_csv(file_path)
    
    # Question 3: Identify top students
    top_students = identify_top_students(students)
    
    print("\nStudents with an average grade of 90 or above:")
    print(f"{'Student_ID':<12}{'Name':<25}{'Average_Grade':<15}")
    print("-" * 50)
    for student_id, name, avg in top_students:
        print(f"{student_id:<12}{name:<25}{avg:.2f}")

if __name__ == "__main__":
    main()
```
