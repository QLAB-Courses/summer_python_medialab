3. Count the Total Trips from JFK for Each Day of the Month

```python
# Display the result
print(trips_from_jfk_per_day)
```

4. Calculate the Average Speed by Pickup Day of the Week

To calculate speed, we use the formula:

\[
\text{Speed} = \frac{\text{Distance}}{\text{Time (in hours)}}
\]

We must convert `trip_length` (seconds) into hours.


```python
# Convert trip_length from seconds to hours
df['trip_length_hours'] = df['trip_length'] / 3600
    
```

```python
# Calculate speed in miles per hour (mph)
df['speed_mph'] = df['trip_distance'] / df['trip_length_hours']
```

```python
# Group by pickup_dayofweek and calculate the mean speed
avg_speed_per_dayofweek = df.groupby('pickup_dayofweek')['speed_mph'].mean()
```

```python
# Display the result
print(avg_speed_per_dayofweek)
```

a. Using a  “for loop” and the conditional “if else” for a range bettewn 1 to 100, evaluate "x" so that:
If x is a multiple of 3 print  “Fizz”
If x is a multiple of 5 print  “Buzz”
If x is a multiple of 3 and 5 print “FizzBuzz”

```python
# Loop from 1 to 100
for x in range(1, 101):  
    # Check if x is divisible by both 3 and 5 first (most restrictive condition)
    if x % 3 == 0 and x % 5 == 0:
        print("FizzBuzz")
    # Check if x is only divisible by 3
    elif x % 3 == 0:
        print("Fizz")
    # Check if x is only divisible by 5
    elif x % 5 == 0:
        print("Buzz")
    # If none of the above, print the number itself
    else:
        print(x)
```

- Accumulate sales totals row by row. Stop the loop when the accumulated revenue exceeds $2000 and print how many sales were necessary.

```python
# Initialize variables
accumulated_revenue = 0
transaction_count = 0
```

```python
# While loop to accumulate revenue per transaction
while accumulated_revenue <= 2000 and transaction_count < len(sales_df):
    accumulated_revenue += sales_df.iloc[transaction_count]["Total"]
    transaction_count += 1

print(f"It took {transaction_count} transactions to exceed $2000.")
```

- Use a combination of `for and while` loops to calculate a rolling cumulative revenue:

For each category, use a for loop to iterate over its transactions.
Use a while loop to stop accumulating revenue for a category once it exceeds $ 3000.

Print the cumulative revenue for each category and the number of transactions it took to exceed $3000.

```python
# Group sales by Category
category_groups = sales_df.groupby("Category")
```

```python
# Loop through each category
for category, group in category_groups:
    accumulated_revenue = 0
    transaction_count = 0
    
    # While loop to accumulate revenue per category
    while accumulated_revenue <= 3000 and transaction_count < len(group):
        accumulated_revenue += group.iloc[transaction_count]["Total"]
        transaction_count += 1

    print(f"{category}: ${accumulated_revenue:.2f} in {transaction_count} transactions")
```

- Use a `nested for` loop to:

Group the sales by year and by date.

This task requires grouping and iterating within groups.

```python
# Convert Date to datetime type
sales_df["Date"] = pd.to_datetime(sales_df["Date"])

# Extract Year
sales_df["Year"] = sales_df["Date"].dt.year

# Group by Year
year_groups = sales_df.groupby("Year")
```

```python
# Iterate through each year
for year, group in year_groups:
    max_revenue_date = group.groupby("Date")["Total"].sum().idxmax()
    max_revenue_value = group.groupby("Date")["Total"].sum().max()
    
    print(f"In {year}, the highest revenue date was {max_revenue_date} with ${max_revenue_value:.2f}")
```

## Part 3  Try, continue and catch exceptions

In this case you will work with a `students_grade.csv` dataset. 

1. You are tasked with calculating the average grade for each student in the dataset. However, the dataset has some issues:

Some scores are marked as "missing" or "invalid" (strings instead of numeric values).
Some students may have incomplete grades across subjects.

To calculate the average:

Skip any "invalid" or "missing" values during calculations.
If a student has no valid grades, print a warning message for that student.


Use `try` and `except` to handle potential errors.
Use `continue` to skip invalid rows or scores.
Print the results in a clear format, showing the average grade for each student or a warning if no grades are available.

```python
import csv
```

```python
#Defining a Function to Calculate Student Average

def calculate_student_average(grades):
    valid_grades = []
    for grade in grades:
        try:
            valid_grades.append(float(grade))  # Convert grade to a float
        except ValueError:
            continue  # Skip invalid data (non-numeric values)
    
    if valid_grades:  # If there are any valid grades
        return sum(valid_grades) / len(valid_grades)
    else:
        return None  # If no valid grades, return None
```

```python
#Defining a Function to Calculate Subject Average

def calculate_subject_average(data, subject_index):
    valid_grades = []
    for row in data:
        try:
            grade = float(row[subject_index])  # Get the grade for this subject
            valid_grades.append(grade)
        except ValueError:
            continue  # Skip invalid data
    
    if valid_grades:
        return sum(valid_grades) / len(valid_grades)
    else:
        return None  # No valid grades, return None
```

```python
# Open the file
with open(r'C:\Users\Michael Encalada\Documents\GitHub\QLAB\summer_python_medialab\assignments\assignment_2\students_grade.csv', mode='r') as file:
    reader = csv.reader(file)
    
    # Skip the header
    headers = next(reader)
    
    student_data = []
    subject_data = {header: [] for header in headers[1:]}  # Exclude student name column

    # Process each student’s grades
    for row in reader:
        student_name = row[0]  # First column is the student's name
        grades = row[1:]  # The rest are the student's grades

        # Calculate student average
        student_average = calculate_student_average(grades)
        if student_average is None:
            print(f"Warning: {student_name} has no valid grades.")
        else:
            print(f"{student_name}: Average grade = {student_average:.2f}")

        student_data.append(grades)

        # Add grades to the subject data for average calculation later
        for index, grade in enumerate(grades):
            subject_data[headers[index + 1]].append(grade)
```

2. Calculate the average grade per subject for each year.
Handle invalid or missing values during the calculation by skipping them.
Print the averages in a clean format (e.g., table or grouped output).

```python
# Calculate and print subject averages
print("\nAverage grade per subject:")
for subject, grades in subject_data.items():
    subject_index = headers.index(subject)  # Get the correct index for the subject
    subject_average = calculate_subject_average(student_data, subject_index)
    if subject_average is None:
        print(f"{subject}: No valid grades available.")
    else:
        print(f"{subject}: Average grade = {subject_average:.2f}")
```

3. Identify students who had an average grade of 90 or above across all years (excluding invalid/missing values).
If a student has no valid grades, skip them with a warning.

```python
# Identifying students with an average >= 90
print("\nStudents with average grade >= 90:")
with open(r'C:\Users\Michael Encalada\Documents\GitHub\QLAB\summer_python_medialab\assignments\assignment_2\students_grade.csv', mode='r') as file:
    reader = csv.reader(file)
    next(reader)  # Skip the header again

    for row in reader:
        student_name = row[0]
        grades = row[1:]

        student_average = calculate_student_average(grades)
        if student_average is not None and student_average >= 90:
            print(f"{student_name}: Average grade = {student_average:.2f}")
```
